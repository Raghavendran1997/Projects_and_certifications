## Gradient Descent – Concept and Intuition

Gradient Descent is an optimization algorithm used to minimize a loss function by iteratively updating model parameters in a direction that reduces prediction error. It is the core learning mechanism behind many machine learning models and neural networks.

---

## Why Gradient Descent Is Required

Most machine learning models define a loss function that measures the difference between predicted and actual values. For complex models, finding the optimal parameters analytically is either infeasible or computationally expensive. Gradient Descent provides an efficient iterative approach to approximate the optimal solution.

---

## Core Intuition

The algorithm works by computing the gradient (slope) of the loss function with respect to the model parameters and updating the parameters in the opposite direction of the gradient.

- A steep slope indicates a large error and results in larger updates.

- A shallow slope indicates smaller updates as the model approaches convergence.

This process continues until the loss stabilizes or reaches a minimum.

---

Mathematical Formulation

The parameter update rule is defined as:

θ = θ - α⋅∇J(θ)

Where:

- θ represents the model parameters (weights)

- α is the learning rate

- ∇J(θ) is the gradient of the loss function with respect to the parameters

---

## Learning Rate (α)

The learning rate controls how large each update step is.

- A large learning rate may cause the model to diverge.

- A small learning rate may slow down convergence.

Choosing an appropriate learning rate is critical for stable and efficient training.

---

## Gradient Descent in Neural Networks

Neural network loss functions are differentiable, allowing gradients to be efficiently computed using backpropagation. Gradient Descent then updates weights layer by layer, enabling the network to learn complex patterns from data.

---

## Practical Considerations

Feature scaling improves convergence speed

Monitoring loss curves helps diagnose training issues

Advanced optimizers such as Momentum, RMSProp, and Adam improve convergence stability

---

## Summary

Gradient Descent is a fundamental optimization technique that enables models to learn from data by iteratively minimizing error through gradient-based parameter updates.

## Gradient Descent for scratch

### Sigmoid Function

```python
def sigmoid(x):
  return 1 / (1 + np.exp(-x))
```

### Prediction Function

```python
def prediction_function(age, affordibility):
  weighted_sum = coef[0]*age + coef[1]*affordibility + intercept
  return sigmoid(weighted_sum)
```

### Loss Function

```python
def log_loss(y_true, y_predicted):
  epsilon = 1e-15
  y_predicted_new = [max(i,epsilon) for i in y_predicted]
  y_predicted_new = [min(i,1-epsilon) for i in y_predicted_new]
  y_predicted_new = np.array(y_predicted_new)

  return -np.mean(y_true * np.log(y_predicted_new) + (1-y_true) * np.log(1-y_predicted_new))
```

## Gradient Descent

```python
def gradient_descent(age, affordibility, y_true, epochs, loss_threshold):
  w1 = w2 = 1
  bias = 0
  rate = 0.5
  n = len(age)

  for i in range(epochs):
    weighted_sum = (w1 * age) + (w2 * affordibility) + bias
    y_predicted = sigmoid(weighted_sum)
    loss = log_loss(y_true, y_predicted)

    w1d = (1/n) * np.dot(np.transpose(age),(y_predicted - y_true))
    w2d = (1/n) * np.dot(np.transpose(affordibility),(y_predicted - y_true))

    bias_d = np.mean(y_predicted - y_true)

    w1 = w1 - rate * w1d
    w2 = w2 - rate * w2d
    bias = bias - rate * bias_d

    print(f'Epochs:{i}, w1:{w1}, w2:{w2}, bias:{bias}, loss:{loss}')

    if loss <= loss_threshold:
      break

    return w1,w2,bias
```

## What the Intercept (Bias) Represents

The intercept value of −0.3970 represents the model’s baseline prediction when all input feature values are zero.

This bias term allows the model to better fit the data by shifting the prediction line (or decision boundary) upward or downward.
