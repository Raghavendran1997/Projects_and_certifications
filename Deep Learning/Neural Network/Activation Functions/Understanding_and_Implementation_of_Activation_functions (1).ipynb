## IMPLEMENTATION OF ACTIVATION FUNCTION IN PYTHON.

## Why Activation Functions Are Needed

Without an activation function, a neural network would behave like a linear
model, regardless of the number of layers. Activation functions allow the
network to learn complex, non-linear relationships in the data.


## Sigmoid Activation Function

The sigmoid function is one of the earliest activation functions used in neural
networks. It introduces non-linearity into the model and maps any real-valued
input to a range between 0 and 1.

This property makes sigmoid especially useful when the output needs to be
interpreted as a probability.

---

## Why Sigmoid?

- Squashes large positive values close to 1 and large negative values close to 0
- Helps model probabilities in binary classification tasks
- Provides smooth gradients, enabling gradient-based optimization

---

## Limitations of Sigmoid

Despite its simplicity, sigmoid has important drawbacks:

- **Vanishing Gradient Problem:**  
  For very large or very small input values, the gradient becomes very close to
  zero, slowing down learning in deep networks.
- **Not Zero-Centered:**  
  Outputs are always positive, which can lead to inefficient weight updates.
- **Slower Convergence:**  
  Compared to ReLU, training can be slower for deeper models.

```python
## SIGMOID
import math

def sigmoid(x):
  return 1 / (1 + math.exp(-x))
```

```python
sigmoid(1)
```

```python
sigmoid(100)
```

```python
sigmoid(-89)
```

## Tanh Activation Function

The tanh (hyperbolic tangent) activation function is a scaled and shifted version
of the sigmoid function. It maps input values to a range between -1 and 1.

Because of this symmetric output, tanh is often preferred over sigmoid in hidden
layers of neural networks.

---

## Why Tanh Is Better Than Sigmoid (in Hidden Layers)

- Outputs are **zero-centered**, which helps balance positive and negative
  gradients during training
- Leads to **faster convergence** compared to sigmoid
- Still introduces non-linearity, allowing the network to learn complex patterns

---

## Mathematical Intuition (High Level)

- Large positive inputs → output close to +1
- Large negative inputs → output close to -1
- Inputs near zero → almost linear behavior

This makes tanh more expressive for feature learning in hidden layers.

---

## Limitations of Tanh

Despite its improvements over sigmoid, tanh also has drawbacks:

- **Vanishing Gradient Problem:**  
  For large input values, gradients still become very small
- Can be sensitive to input scaling
- Not ideal for very deep networks




```python
def tanh(x):
  return (math.exp(x) - math.exp(-x)) / (math.exp(x) + math.exp(-x))
```

```python
tanh(-56)
```

```python
tanh(56)
```

```python
tanh(1)
```

## ReLU Activation Function

ReLU (Rectified Linear Unit) is one of the most widely used activation functions
in modern neural networks. It introduces non-linearity by allowing only positive
values to pass through while setting all negative values to zero.

ReLU is defined as:
f(x) = max(0, x)

---

## Why ReLU Became the Default Choice

ReLU addresses many of the limitations of sigmoid and tanh:

- **Prevents Vanishing Gradients:**  
  For positive input values, gradients remain strong, enabling faster learning
- **Computationally Efficient:**  
  Simple mathematical operation, making training faster
- **Sparse Activation:**  
  Many neurons output zero, improving efficiency and reducing overfitting

---

## Intuition

- Negative input → output is 0 (neuron inactive)
- Positive input → output grows linearly (neuron active)

This behavior allows the network to focus on important features while ignoring
irrelevant ones.

---

## Limitations of ReLU

Despite its advantages, ReLU has its own challenges:

- **Dying ReLU Problem:**  
  Neurons can become permanently inactive if they output zero for all inputs
- Output is not zero-centered
- Can be sensitive to large learning rates




```python
def relu(x):
  return max(0,x)
```

```python
relu(-9)
```

```python
relu(9)
```

## Leaky ReLU Activation Function

Leaky ReLU is a variant of the ReLU activation function designed to address the
"Dying ReLU" problem. Unlike ReLU, which outputs zero for all negative inputs,
Leaky ReLU allows a small, non-zero gradient when the input is negative.

This helps ensure that neurons continue to learn even when they receive negative
input values.

Leaky ReLU is defined as:
f(x) = x            if x > 0
f(x) = αx           if x ≤ 0

where α is a small constant (commonly 0.01).

---

## Why Leaky ReLU Was Introduced

ReLU neurons can sometimes become inactive and stop learning entirely if they
consistently output zero. Leaky ReLU solves this by:

- Allowing gradients to flow even for negative inputs
- Reducing the risk of neurons dying permanently
- Improving learning stability in deeper networks

---

## Intuition

- Positive input → output increases linearly (same as ReLU)
- Negative input → output decreases slowly instead of becoming zero

This ensures that all neurons remain at least partially active during training.

---

## Advantages Over ReLU

- Prevents the dying ReLU problem
- Maintains non-linearity
- Simple and computationally efficient

---

## Limitations

- The negative slope α is a hyperparameter that must be chosen manually
- Performance gains over ReLU are sometimes dataset-dependent
- Still not zero-centered




```python
def leaky_relu(x):
  return max(0.1*x,x)
```

```python
leaky_relu(-100)
```

```python
leaky_relu(67)
```

## Activation Functions Comparison

| Activation Function | Output Range | Key Advantage | Main Limitation | Common Usage |
|---------------------|--------------|---------------|-----------------|--------------|
| Sigmoid | (0, 1) | Interpretable as probability | Vanishing gradient, slow training | Output layer (binary classification) |
| Tanh | (-1, 1) | Zero-centered, faster convergence | Vanishing gradient | Hidden layers (older architectures) |
| ReLU | [0, ∞) | Fast training, stable gradients | Dying ReLU problem | Hidden layers (default choice) |
| Leaky ReLU | (-∞, ∞) | Prevents dying ReLU | Extra hyperparameter (α) | Hidden layers (when ReLU underperforms) |

---

## Key Takeaway

Choosing the right activation function has a significant impact on how efficiently
a neural network learns. While sigmoid and tanh helped establish early neural
networks, ReLU and its variants are better suited for deeper models due to more
stable gradient flow.
